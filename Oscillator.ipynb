The goal of this simulation is to computationally solve the governing differential equations of a mechanical oscillator over a 20-second time domain. By utilizing numerical integration (`scipy.integrate.odeint`), we can model the transient dynamic response of the system based on its known physical parameters.Governing Physics & State-Space Formulation.


The movement of a simple harmonic oscillator is governed by Newton's Second Law, which yields a classic second-order Ordinary Differential Equation (ODE):$$m \frac{d^2x}{dt^2} + c \frac{dx}{dt} + kx = 0$$

Where:
* $m$ = Mass (inertia)
* $c$ = Damping coefficient (energy dissipation)
* $k$ = Spring constant (stiffness)
* $x$ = Position
* $t$ = Time

Because computational ODE solvers strictly require first-order differential equations, we must reduce this second-order ODE into a system of two first-order ODEs (State-Space formulation). By defining velocity as $v = \frac{dx}{dt}$, we get:
$$\frac{dx}{dt} = v$$
$$\frac{dv}{dt} = \frac{-cv - kx}{m}$$

```python
from scipy.integrate import odeint
import numpy as np
import matplotlib.pyplot as plt
```

```python
def msd_system(state, t, m, c, k):
    dxdt = state[1]
    dvdt = (-c * state[1] - k * state[0]) / m
    return [dxdt, dvdt]
```

```python

print("=== Define parameters for the mass-spring-damper system ===\n")
m=float(input("Enter mass (m): "))
c=float(input("Enter damping coefficient (c): "))
k=float(input("Enter spring constant (k): "))

print("\n=== Define initial conditions ===\n")
time_array=np.linspace(0, 20, 200)  # Time from 0 to 10 seconds
x, v = (float(input("Enter initial position (x0): ")), float(input("Enter initial velocity (v0): ")))
initial_state= [x,v]

print("Given initial conditions:")
print(f"---> Initial position (x0): {x}")
print(f"---> Initial velocity (v0): {v}\n")  
print("\nGiven system parameters:")
print(f"---> Mass (m): {m}")
print(f"---> Damping coefficient (c): {c}")
print(f"---> Spring constant (k): {k}\n")
solution = odeint(msd_system, initial_state, time_array, args=(m, c, k))
```

To validate the physical realism of the model, the script mathematically classifies the nature of the system's damping . The critical damping threshold ($c_c$) is calculated using the system's mass and stiffness:
$$c_c = 2\sqrt{mk}$$

The system is then classified conditionally:
* Underdamped: $c < c_c$ (The mass oscillates back and forth, slowly decaying)
* Critically Damped: $c = c_c$ (The mass returns to equilibrium as fast as possible without oscillating)
* Overdamped: $c > c_c$ (The mass returns to equilibrium slowly without oscillating)

```python
x_all = solution.T[0]  # Position over time
v_all = solution.T[1]  # Velocity over time
print("\n==================== Nature of Damping of System ====================\n")

damp_critical = 2 * np.sqrt(m * k)
print(f"----> Critical damping coefficient (c_critical): {damp_critical}")
print(f"----> Actual damping coefficient (c): {c}")
print()

if c < damp_critical:
    print("Damping Nature ==> The system is underdamped.")
elif c == damp_critical:
    print("Damping Nature ==> The system is critically damped.")
else:
    print("Damping Nature ==> The system is overdamped.")
```

```python

print("\n==================== Simulation Results ====================\n")

x_t=plt.figure(figsize=(12, 6))
plot1=x_t.add_subplot(1, 1, 1)
plot1.plot(time_array, x_all, label='Position (x)', color='blue')
plot1.set_title('Variation of Position with Time')
plot1.set_xlabel('Time (s)')
plot1.set_ylabel('Position (m)')
plot1.legend()
plot1.grid(True)

v_t=plt.figure(figsize=(12, 6))
plot2=v_t.add_subplot(1, 1, 1)
plot2.plot(time_array, v_all, label='Velocity (v)', color='red')
plot2.set_title('Variation of Velocity with Time')
plot2.set_xlabel('Time (s)')
plot2.set_ylabel('Velocity (m/s)')
plot2.legend()
plot2.grid(True)

plt.tight_layout()
```
